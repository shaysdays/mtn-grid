```python
import pandas as pd
from mtn_grid.storage import load_processed_activities
from mtn_grid.summits import add_summit_metrics

df = load_processed_activities()
df.shape
```

```python
# Add summit metrics
df = add_summit_metrics(df, 'sanitas', 40.0344166667, -105.30525, enter_m=80, exit_m=120, exit_consec_points=5)
df = add_summit_metrics(df, 'green', 39.98215812792686, -105.30158524126522, enter_m=80, exit_m=120, exit_consec_points=5)

# Summaries
sanitas_touches = int(df['sanitas_touches'].sum())
sanitas_touch_activities = int(df['sanitas_summit'].sum())
sanitas_days = df.loc[df['sanitas_summit'], 'start_date_local'].dt.date.nunique()

green_touches = int(df['green_touches'].sum())
green_touch_activities = int(df['green_summit'].sum())
green_days = df.loc[df['green_summit'], 'start_date_local'].dt.date.nunique()

print(f'Sanitas summits: {sanitas_touches}')
print(f'Activities that hit at least 1 Sanitas summit: {sanitas_touch_activities}')
print(f'Unique Sanitas summit days: {sanitas_days}\n')
print(f'Green summits: {green_touches}')
print(f'Activities that hit at least 1 Green summit: {green_touch_activities}')
print(f'Unique Green summit days: {green_days}')
```

```python
# Multi days Sanitas

multi_sanitas = (
    df.loc[df['sanitas_touches'] >= 2,
                          ['start_date_local', 'name', 'sport_type', 'sanitas_touches',
                           'distance_mi', 'elevation_gain_ft']]
    .sort_values('start_date_local')
)

print('\nActivities with 2+ Sanitas touches:')
print(multi_sanitas.to_string(index=False))
```

```python
# Multi days Green

multi_green = (
    df.loc[df['green_touches'] >= 2,
                          ['start_date_local', 'name', 'sport_type', 'green_touches',
                           'distance_mi', 'elevation_gain_ft']]
    .sort_values('start_date_local')
)

print('\nActivities with 2+ Green touches:')
print(multi_green.to_string(index=False))
```
